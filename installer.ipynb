```python
# ============================================================
# SKETCHBYTE — QWEN3 TTS VOICE SYSTEM
# CELL 1 — SETUP & ENVIRONMENT
# ============================================================

import sys
import gc
import subprocess

if "_sb_error" not in globals():
    def _sb_error(title, detail, hint=None):
        lines = [
            "=" * 70,
            "❌ SKETCHBYTE ERROR",
            "=" * 70,
            "",
            str(title),
        ]
        detail_text = str(detail).strip()
        if detail_text:
            lines += ["", detail_text]
        if hint:
            lines += ["", f"💡 HINT: {hint}"]
        return RuntimeError("\n".join(lines))

print("=" * 70)
print("SKETCHBYTE — QWEN3 TTS VOICE SYSTEM")
print("CELL 1 — SETUP & ENVIRONMENT")
print("=" * 70)


# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

print("\n[1/4] Installing required packages...")
print("-" * 70)

packages = [
    "qwen-tts",
    "transformers",
    "accelerate",
    "soundfile",
    "librosa",
    "ipywidgets"
]

try:
    subprocess.check_call(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "-U",
            *packages
        ]
    )

    print("✅ Required packages installed.")

except Exception as e:
    raise _sb_error(
        "PACKAGE INSTALLATION FAILED",
        str(e),
        "Restart the runtime, then re-run Cell 1. Check your network connection if it persists.",
    )


# ============================================================
# 2. IMPORT CORE LIBRARIES
# ============================================================

print("\n[2/4] Loading Python libraries...")
print("-" * 70)

try:
    import torch
    import soundfile as sf

    print("✅ PyTorch loaded.")
    print("✅ SoundFile loaded.")

except Exception as e:
    raise _sb_error(
        "LIBRARY IMPORT FAILED",
        str(e),
        "Restart the Colab runtime and run Cell 1 again.",
    )


# ------------------------------------------------------------
# VERIFY ALL REQUIRED PACKAGES ARE IMPORTABLE
# ------------------------------------------------------------

missing_modules = []
for module_name in [
    "transformers",
    "accelerate",
    "librosa",
    "ipywidgets",
    "qwen_tts",
]:
    try:
        __import__(module_name)
    except Exception:
        missing_modules.append(module_name)

if missing_modules:
    raise _sb_error(
        "MISSING PYTHON PACKAGES",
        "Could not import: " + ", ".join(missing_modules),
        "Run Runtime > Restart session, then re-run Cell 1. Check your network if it persists.",
    )

print("✅ All required packages verified.")


# ============================================================
# 3. GPU / CUDA CHECK
# ============================================================

print("\n[3/4] Checking GPU...")
print("-" * 70)

if not torch.cuda.is_available():

    print("❌ NO GPU DETECTED")

    raise RuntimeError(
        "\n"
        "====================================================\n"
        "❌ SKETCHBYTE REQUIRES A GPU\n"
        "====================================================\n\n"
        "Go to:\n"
        "Runtime → Change runtime type → T4 GPU\n\n"
        "Then reconnect the runtime and run Cell 1 again.\n"
    )

gpu_name = torch.cuda.get_device_name(0)

gpu_properties = torch.cuda.get_device_properties(0)

gpu_vram = gpu_properties.total_memory / (1024 ** 3)

print(f"✅ GPU: {gpu_name}")
print(f"✅ CUDA: {torch.version.cuda}")
print(f"✅ PyTorch: {torch.__version__}")
print(f"✅ VRAM: {gpu_vram:.1f} GB")


# ============================================================
# 4. SKETCHBYTE CONFIGURATION
# ============================================================

print("\n[4/4] Creating SketchByte configuration...")
print("-" * 70)


# ------------------------------------------------------------
# MODELS
# ------------------------------------------------------------

FORMATTER_MODEL = "Qwen/Qwen3-4B-Instruct-2507"

TTS_MODEL = "Qwen/Qwen3-TTS-12Hz-1.7B-Base"


# ------------------------------------------------------------
# PERMANENT VOICE REFERENCE
# ------------------------------------------------------------

# Reference voice file — must live at: My Drive/SketchByte/Voice/fishengagement.mp3
REFERENCE_AUDIO = "fishengagement1.mp3"

REFERENCE_TEXT = """
He was, by any measure, one of the smartest men who ever lived.
He reshaped our understanding of gravity, light, and motion itself.
Universities still teach his laws four centuries later.

And in the summer of seventeen twenty, this same man looked at his investment.
""".strip()


# ------------------------------------------------------------
# OUTPUT
# ------------------------------------------------------------

OUTPUT_FILE = "SketchByte_Voiceover.wav"


# ------------------------------------------------------------
# GLOBAL VARIABLES
# These will be filled by later cells.
# ------------------------------------------------------------

formatter_model = None
formatter_tokenizer = None

tts_model = None

# ============================================================
# FINAL SYSTEM STATUS
# ============================================================

print("\n" + "=" * 70)
print("SKETCHBYTE CONFIGURATION")
print("=" * 70)

print(f"\n🧠 Script Formatter:")
print(f"   {FORMATTER_MODEL}")

print(f"\n🎙️ Voice Model:")
print(f"   {TTS_MODEL}")

print(f"\n🎧 Permanent Reference:")
print(f"   {REFERENCE_AUDIO}")

print(f"\n📄 Reference Transcript:")
print(f"   {len(REFERENCE_TEXT.split())} words")

print(f"\n📁 Output:")
print(f"   {OUTPUT_FILE}")


# ============================================================
# MEMORY CLEANUP
# ============================================================

gc.collect()
torch.cuda.empty_cache()


# ============================================================
# READY
# ============================================================

print("\n" + "=" * 70)
print("🟢 CELL 1 COMPLETE")
print("=" * 70)

print("\nEnvironment      : ✅")
print("Packages         : ✅")
print("GPU              : ✅")
print("CUDA             : ✅")
print("Configuration    : ✅")

print("\n➡️ Next: Run CELL 2")
print("CELL 2 will prepare the permanent reference voice")
print("and load the Qwen3-4B script formatter.")
print("=" * 70)
```
